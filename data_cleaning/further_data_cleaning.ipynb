```python
def preprocess_data(df):
    # 处理Levy列
    df['Levy'] = df['Levy'].replace('-', np.nan).astype(float)
    
    # 处理Doors列：提取有效数字并转换
    def parse_doors(val):
        # 处理特殊符号如">5"
        if isinstance(val, str) and '>' in val:
            return int(val.replace('>', ''))
        # 处理日期格式如"04-May"
        if isinstance(val, str) and '-' in val:
            return int(val.split('-')[0])
        # 常规数字处理
        try:
            return int(val)
        except:
            return np.nan  # 无法解析的设为缺失值
    
    df['Doors'] = df['Doors'].apply(parse_doors).astype('Int64')
    
    # 处理其他列
    df['Leather interior'] = df['Leather interior'].map({'Yes': 1, 'No': 0})
    df = df.drop(['ID'], axis=1)
    
    return df
```

```python
# 创建组合特征
df['Age'] = 2025 - df['Prod. year']  # 车龄
df['Miles_per_year'] = df['Mileage'] / df['Age']  # 年均里程
df['Brand_Model'] = df['Manufacturer'] + '_' + df['Model'] # 品牌-车型组合特征
```

```python
df = preprocess_data(df)
```
