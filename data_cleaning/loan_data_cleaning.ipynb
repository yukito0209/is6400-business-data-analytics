```python
loan_status = df['loan_status'].value_counts(dropna=False)
loan_status
```

```python
# 年龄超过100岁的数据不合理
df = df[df['person_age'] <= 100]
```

```python
# 处理极高收入（使用IQR方法）
Q1 = df['person_income'].quantile(0.25)
Q3 = df['person_income'].quantile(0.75)
IQR = Q3 - Q1
upper_income = Q3 + 1.5 * IQR
df = df[df['person_income'] <= upper_income]
```

```python
# 处理不合理的工作经验（最大60年）
df = df[df['person_emp_exp'] <= 60]
```

```python
# 转换数据类型
df['person_age'] = df['person_age'].astype(int)
df['cb_person_cred_hist_length'] = df['cb_person_cred_hist_length'].astype(int)
```
